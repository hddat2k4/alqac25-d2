# Step 1: Chunk data

```python
from utils import load_json, rechunk_data
```

```python
chunk_threshold = 500
```

```python
law_data = load_json('data/alqac25_law.json')
```

```python
rechunked_data = rechunk_data(data=law_data, threshold=chunk_threshold, output_file=f'data/private/alqac25_law_rechunked_{chunk_threshold}.json')
```

# Step 2: Embedding data

```python
from utils import data_to_dict, embed_data, create_hybrid_collection, insert_data_with_vectors
```

```python
data_for_embedding = data_to_dict(rechunked_data)
```

```python
embedding_data = embed_data(data=data_for_embedding, output_file='data/private/embeddings.json')
```

```python
name = f"alqac_law_collection_chunk_{chunk_threshold}"
```

```python
create_hybrid_collection(name)
```

```python
insert_data_with_vectors(name, embedding_data)
```

# Step 3: Retrieve 

```python
from utils import retrieve
```

```python
retrieve(input_file="data/private/alqac25_private_test_Task_1.json",
        output_file="data/private/retrieval_candidates_rechunked_500_PRIVATE.json",
        name=name,
        mode="hybrid",
        top_k=12,
        alpha=0.8)
```
